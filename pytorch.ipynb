```python
import torch

# Scalar
x = torch.tensor(5)

# Vector
y = torch.tensor([1, 2, 3])

# Matrix
z = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(x)
print(y)
print(z)
```

```python
print(x.shape)
print(y.shape)
print(z.shape)
```

```python
print(z.shape)
print(z.ndim)
print(z.dtype)
print(z.device)
```

```python
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
a = torch.tensor([
    [1.0, 2.0],
    [3.0, 4.0]
])

b = torch.tensor([
    [5.0, 6.0],
    [7.0, 8.0]
])

result = torch.matmul(a, b)

print(result)
```

```python
x = torch.tensor([1, 2, 3, 4, 5, 6])

print(x.shape)
```

```python
x = x.reshape(2, 3)

print(x)
print(x.shape)
```

```python
x = torch.zeros(3, 4)  # Matrix filled with zeros

y = torch.ones(3, 4)   # Matrix filled with ones

z = torch.randn(3, 4)  # Random numbers

a = torch.arange(10)   # Numbers from 0 to 9

print(x)
print(y)
print(z)
print(a)
```

```python
import torch

print("PyTorch version:", torch.__version__)

print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Running on CPU")
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
```

```python
x = torch.tensor([1.0, 2.0, 3.0])

x = x.to(device)

print(x.device)
```

```python
print(torch.cuda.memory_allocated() / 1024**2, "MB")
print(torch.cuda.memory_reserved() / 1024**2, "MB")
```
